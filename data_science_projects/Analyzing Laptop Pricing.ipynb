# **Data Wrangling**
<a id= "wrangling"></a>

</div>

In this section, I handled missing data, corrected the data type of the dataframe attribute and executed the processes of data standardization and data normalization on specific attributes of the dataset.

### Objectives

 - Handle missing data in different ways
 - Correct the data type of different data values as per requirement
 - Standardize and normalize the appropriate data attributes
 - Visualize the data as grouped bar graph using Binning
 - Converting a categorical data into numerical indicator variables

````python
import warnings
warnings.filterwarnings("ignore")

import matplotlib.pyplot as plt
%matplotlib inline
````

Updating the `Screen_Size_inch` column  rounded to nearest 2 decimal places by using `numpy.round()` in cm

````python
df[['Screen_Size_cm']] = df[['Screen_Size_cm']].astype(float)
df['Screen_Size_cm'] = np.round(df[['Screen_Size_cm']],2)
df.head()
````

Identifing the entries having Null values.

````python
missing_data = df.isnull()
print(missing_data.head())
for column in missing_data.columns.values.tolist():
    print(column)
    print (missing_data[column].value_counts())
    print("")  
````

Missing values in attributes that have continuous data are best replaced using mean value. Values in "Weight_kg" attribute are continuous in nature, and some values are missing. Therefore, replacing the missing values of weight with the average value of the attribute.

````python
avg_weight= df["Weight_kg"].astype("float").mean().round(2)
df["Weight_kg"].replace(np.nan, avg_weight, inplace=True)
df["Weight_kg"].info();
````

Missing values in attributes that have categorical data are best replaced using the most frequent value. Values in "Screen_Size_cm" attribute are categorical in nature, and some values are missing. Therefore, replacing the missing values of Screen Size with the most frequent value of the attribute.

````python
freq_screensize= df["Screen_Size_cm"].value_counts().idxmax()
df["Screen_Size_cm"].replace(np.nan, freq_screensize, inplace=True)
df["Screen_Size_cm"].info()
````

Both "Weight_kg" and "Screen_Size_cm" are seen to have the data type "Object", while both of them should be having a data type of "float".

````python
df[["Weight_kg", "Screen_Size_cm"]]= df[["Weight_kg", "Screen_Size_cm"]].astype("float")
````

### Data Standardization
The value of Screen_size usually has a standard unit of inches. Similarly, weight of the laptop is needed to be in pounds.

```{math}
1 inch = 2.54 cm
1 kg   = 2.205 pounds
```


````python
df["Screen_Size_cm"]=np.round(df["Screen_Size_cm"]/2.54, 2)
df["Weight_kg"]= np.round(df["Weight_kg"]*2.205, 2)
df.rename(columns={"Screen_Size_cm":"Screen_Size_inch", 'Weight_kg':'Weight_pounds'}, inplace=True)
df.head()
````

### Data Normalization
Normalizing a continuous CPU frequency attribute.


df["CPU_frequency"]= df["CPU_frequency"]/df["CPU_frequency"].max()
df.he

````python
df["CPU_frequency"]= df["CPU_frequency"]/df["CPU_frequency"].max()
df.head()
````

### Binning
Binning is a process of creating a categorical attribute which splits the values of a continuous data into a specified number of groups. In this case, I created 3 bins for the attribute `Price`. These bins were named `"Low", "Medium", "High"`. The new attribute will be named `Price-binned`.

````python
bins = np.linspace(min(df["Price"]), max(df["Price"]), 4)
group_names = ['Low', 'Medium', 'High']
df['Price-binned'] = pd.cut(df['Price'], bins, labels=group_names, include_lowest=True )
df.head()
````

````python
df["Price-binned"].value_counts()
````

Ploting the bar graph of these bins.

````python
from matplotlib import pyplot
plt.bar(group_names, df["Price-binned"].value_counts())

plt.xlabel("Price")
plt.ylabel("Count")
plt.title("Price for different Laptops")
````

Converting the `Screen` attribute of the dataset into 2 indicator variables, `Screen-IPS_panel` and `Screen-Full_HD`. Then droping the `Screen` attribute from the dataset.

````python
dummy_screen= pd.get_dummies(df["Screen"])
dummy_screen.rename(columns={"IPS Panel":"Screen-IPS_panel", "Full HD":"Screen-Full_HD"}, inplace=True)
df= pd.concat([df, dummy_screen], axis=1)
df.drop("Screen", axis=1, inplace= True)
df.head()
````

````python
df.to_csv("cleaned_laptop.csv")
````

# **Machine Learning Model Development**

<a id="machine"></a>

</div>

In this section, I used linear regression principles to create a model that predicts the price of the laptop, based on one or more attributes of the dataset.

# Objectives

 - Use Linear Regression in one variable to fit the parameters to a model
 - Use Linear Regression in multiple variables to fit the parameters to a model
 - Use Polynomial Regression in single variable tofit the parameters to a model
 - Create a pipeline for performing linear regression using multiple features in polynomial scaling
 - Evaluate the performance of different forms of regression on basis of MSE and R^2 parameters

````python
!pip install -q scikit-learn --disable-pip-version-check
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, r2_score
import warnings
warnings.filterwarnings("ignore", category=UserWarning) 
````

### Single Linear Regression

`CPU_frequency` is the parameter with the lowest p-value among the different features of the dataset. Creating a single feature Linear Regression model that fits the pair of `CPU_frequency` and `Price` to find the model for prediction.

````python
lm= LinearRegression()
X= df[["CPU_frequency"]]
Y= df["Price"]
lm.fit(X, Y)
Yhat= lm.predict(X)
print(Yhat[0:10])
````

Creating a Distribution plot for the predicted values and that of the actual values.

````python
ax1 = sns.distplot(df['Price'], hist=False, color="r", label="Actual Value")
sns.distplot(Yhat, hist=False, color="b", label="Fitted Values" , ax=ax1)

plt.title('Actual vs Fitted Values for Price')
plt.xlabel('Price')
plt.ylabel('Proportion of laptops')
plt.legend(['Actual Value', 'Predicted Value'])
plt.show();
````

Evaluating the Mean Squared Error and R^2 score values for the model. 


````python
mse_slr = mean_squared_error(df['Price'], Yhat)
r2_score_slr = lm.score(X, Y)
print('The R-square for Linear Regression is: ', round(r2_score_slr, 3))
print('The mean square error of price and predicted value is: ', round(mse_slr, 3))
````

### Multiple Linear Regression
The parameters which have a low enough p-value so as to indicate strong relationship with the `Price` value are `CPU_frequency`, `RAM_GB`, `Storage_GB_SSD`, `CPU_core`, `OS`, `GPU` and `Category`. Using all these variables, I created a Multiple Linear Regression system. 

````python
lm1= LinearRegression()
Z= df[['CPU_frequency','RAM_GB','Storage_GB_SSD','CPU_core','OS','GPU','Category']]
lm1.fit(Z, Y)
Y_hat= lm1.predict(Z)
Y_hat[0:5]
````

Ploting the Distribution graph of the predicted values as well as the Actual values


````python
ax1 = sns.distplot(df['Price'], hist=False, color="r", label="Actual Value")
sns.distplot(Y_hat, hist=False, color="b", label="Fitted Values" , ax=ax1)

plt.title('Actual vs Fitted Values for Price')
plt.xlabel('Price')
plt.ylabel('Proportion of laptops')
plt.legend(['Actual Value', 'Predicted Value'])
plt.show()
````

Checking the R^2 score and the MSE value for this fit to know whether it is better or worst than the performance of Single Linear Regression.

````python
mse_slr = mean_squared_error(df['Price'], Y_hat)
r2_score_slr = lm1.score(Z, Y) 
print('The R-square for Linear Regression is: ', round(r2_score_slr, 3))
print('The mean square error of price and predicted value is: ', round(mse_slr, 3))
````

### Polynomial Regression
Using the variable "CPU_frequency" to create Polynomial features.


````python
X = X.flatten()
feature_1 = np.polyfit(X, Y, 1)
poly_1 = np.poly1d(feature_1)

feature_3 = np.polyfit(X, Y, 3)
poly_3 = np.poly1d(feature_3)

feature_5 = np.polyfit(X, Y, 5)
poly_5 = np.poly1d(feature_5)
````

Ploting the regression output against the actual data points to note how the data fits in each case.

````python
def PlotPolly(model, independent_variable, dependent_variabble, Name):
    x_new = np.linspace(independent_variable.min(),independent_variable.max(),100)
    y_new = model(x_new)

    plt.plot(independent_variable, dependent_variabble, '.', x_new, y_new, '-')
    plt.title(f'Polynomial Fit for Price vs CPU frequency')
    ax = plt.gca()
    ax.set_facecolor((0.898, 0.898, 0.898))
    fig = plt.gcf()
    plt.xlabel("CPU frequency")
    plt.ylabel('Price of laptops')
    
PlotPolly(poly_1, X, Y, 'CPU_frequency')
````

````python
PlotPolly(poly_3, X, Y, "CPU_Frequency")
````

````python
PlotPolly(poly_5, X, Y, "CPU_Frequency")
````

Also, calculating the `R^2` and `MSE` values for these fits. For polynomial functions, the function `sklearn.metrics.r2_score` was used to calculate R^2 values.

````python
r_squared_1 = r2_score(Y, poly_1(X))
print('The R-square value for 1st degree polynomial is: ', round(r_squared_1, 3))
print('The MSE value for 1st degree polynomial is: ', round(mean_squared_error(Y,poly_1(X)), 3), "\n")
r_squared_3 = r2_score(Y, poly_3(X))
print('The R-square value for 3rd degree polynomial is: ', round(r_squared_3, 3))
print('The MSE value for 3rd degree polynomial is: ',round(mean_squared_error(Y,poly_3(X)), 3), "\n")
r_squared_5 = r2_score(Y, poly_5(X))
print('The R-square value for 5th degree polynomial is: ', round(r_squared_5, 3))
print('The MSE value for 5th degree polynomial is: ', round(mean_squared_error(Y,poly_5(X)), 3), "\n")
````

### Pipeline
Creating a pipeline that performs parameter scaling, Polynomial Feature generation and Linear regression. Using the set of multiple features as before to create this pipeline. 

````python
input=[("scale", StandardScaler()), ("polynomial", PolynomialFeatures(include_bias=False)), ("model", LinearRegression())]
pipe= Pipeline(input)
Z= Z.astype(float)
pipe.fit(Z, Y)
ypipe=pipe.predict(Z)
````

Evaluating the MSE and R^2 values for the this predicted output.


````python
print('MSE for multi-variable polynomial pipeline is: ', round(mean_squared_error(Y, ypipe), 3))
print('R^2 for multi-variable polynomial pipeline is: ', round(r2_score(Y, ypipe), 3))
````

As we can see, values of R^2 increase as I went from Single Linear Regression to Multiple Linear Regression. Further, if we go for multiple linear regression extended with polynomial features, I got an even better R^2 value.

# **Model Evaluation and Refinement**

<a id= "model"></a>
</div>

In this section, I tried to refine the model's performance in predicting the price of a laptop, given the attribute values.

### Objectives

 - Use training, testing and cross validation to improve the performance of the dataset. 
 - Identify the point of overfitting of a model
 - Use Ridge Regression to identify the change in performance of a model based on its hyperparameters
 - Use Grid Search to identify the best performing model using different hyperparameters


````python
!pip install -q tqdm --disable-pip-version-check
from tqdm import tqdm
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.preprocessing import PolynomialFeatures;
````

### Using Cross validation to improve the model

Dividing the dataset into x_data and y_data parameters. Here y_data is the "Price" attribute, and x_data has all other attributes in the data set.

````python
y_data=df["Price"]
x_data=df.drop("Price", axis=1)
````

Spliting the data set into training and testing subests with 10% of the data set for testing purposes.

````python
x_train, x_test, y_train, y_test= train_test_split(x_data, y_data, test_size=0.10, random_state=1)
print("Number of test samples:", x_test.shape[0])
print("Number of training samples:", x_train.shape[0])
````

Creating a single variable linear regression model using "CPU_frequency" parameter.


````python
lr= LinearRegression()
lr.fit(x_train[["CPU_frequency"]], y_train)
print(f"R^2 value for training set: {round(lr.score(x_train[['CPU_frequency']], y_train), 3)}")
print(f"R^2 value for tesing set: {round(lr.score(x_test[['CPU_frequency']], y_test), 3)}")
````

Runing a 4-fold cross validation on the model and printing the mean value of R^2 score along with its standard deviation.

````python
cross= cross_val_score(lr, x_data[["CPU_frequency"]], y_data, cv=4)
print(f"R^2 score: {round(cross.mean(), 3)}")
print(f"Standard deviation: {round(cross.std(), 3)}")
````

### Overfitting

Spliting the data set into training and testing components again but this time reserving 50% of the data set for testing.

````python
x_train1, x_test1, y_train1, y_test1= train_test_split(x_data, y_data, test_size=0.50, random_state=0)
````

To identify the point of overfitting the model on the parameter "CPU_frequency", I created polynomial features using the single attribute. 

Then, I evaluated the R^2 scores of the model created using different degrees of polynomial features, ranging from 1 to 5.

````python
R2_test= []
order=[1,2,3,4,5]
for n in order:
    pr= PolynomialFeatures(degree=n)
    x_train1_pr= pr.fit_transform(x_train1[["CPU_frequency"]])
    x_test1_pr= pr.fit_transform(x_test1[["CPU_frequency"]])
    lr.fit(x_train1_pr, y_train1)
    R2_test.append(lr.score(x_test1_pr, y_test1))

for order_2, R2 in zip(order, R2_test):
    print(f"Degree: {order_2}, R^2: {round(R2, 3)}")
````

Ploting the values of R^2 scores against the order.

````python
plt.plot(order, R2_test)
plt.xlabel('order')
plt.ylabel('R^2')
plt.title('R^2 Using Test Data');
````

### Ridge Regression

Creating a polynomial feature model that uses all parameters with degree=2. 


````python
pr = PolynomialFeatures(degree=2)
x_train1_pr = pr.fit_transform(x_train1[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']])
x_test1_pr = pr.transform(x_test1[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']])
````

Creating a Ridge Regression model and evaluating it using values of the hyperparameter alpha ranging from 0.001 to 1 with increments of 0.001. 

Then, creating a list of all Ridge Regression R^2 scores for training and testing data. 


````python
R2_test=[]
R2_train=[]
Alpha = np.arange(0.001,1,0.001)
pbar= tqdm(Alpha)

for alpha in pbar:
    RidgeModel= Ridge(alpha=alpha)
    RidgeModel.fit(x_train1_pr, y_train1)
    test_score, train_score= RidgeModel.score(x_test1_pr, y_test1), RidgeModel.score(x_train1_pr, y_train1)
    pbar.set_postfix({"Test Score": test_score, "Train Score": train_score})
    R2_test.append(test_score)
    R2_train.append(train_score)
````

Ploting the R^2 values for training and testing sets with respect to the value of alpha

````python
plt.figure(figsize=(10, 6))  
plt.plot(Alpha, R2_test, label='validation data')
plt.plot(Alpha, R2_train, 'r', label='training Data')
plt.xlabel('alpha')
plt.ylabel('R^2')
plt.ylim(0, 1)
plt.legend();
````

### Grid Search

Using the raw data and the same set of features as used above, I used GridSearchCV to identify the value of alpha for which the model performs best. 

```math
{0.0001,  0.001,  0.01,  0.1,  1,  10}
```

Creating a Ridge instance and running Grid Search using a 4 fold cross validation.


````python
parameters1= [{'alpha': [0.0001,0.001,0.01, 0.1, 1, 10]}]
RR=Ridge()
Grid1 = GridSearchCV(RR, parameters1,cv=4)
Grid1.fit(x_train[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core', 'OS', 'GPU', 'Category']], y_train)
````

Printing the R^2 score for the test data using the estimator that uses the derived optimum value of alpha.


````python
BestRR=Grid1.best_estimator_
print(round(BestRR.score(x_test[['CPU_frequency', 'RAM_GB', 'Storage_GB_SSD', 'CPU_core','OS','GPU','Category']], y_test), 3))
````
