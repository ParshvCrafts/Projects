### Monthly Sales Analysis

For monthly sales analysis, I grouped the data by the "Order_Month" column and calculated the sum of the "Sales" column for each month. Then, I sorted the results in descending order and reset the index.

```python
sales_by_month = df.groupby("Order_Month")["Sales"].sum().reset_index()
sales_by_month.sort_values(by="Sales", ascending=False, inplace=True)
sales_by_month.reset_index(drop=True, inplace=True)
sales_by_month
```

November is the month with the highest sales of 352461 units, followed by December and September. However, February and January have the lowest sales.

Creating a line chart to visualize the sales trend over time.

```python
fig = px.line(sales_by_month.sort_values('Order_Month'), x='Order_Month', y='Sales', title='Sales by Month')
fig.show()
```

Line chart makes it easy to see the trend of sales over time. As we can see that sales increase from February (59.7k) to March (205k) and then drops a bit in April (138k). The sales remain stable from May to August at 150k. Then, sales largley increases in September (307k) and then drops in October (200k). However, it reaches the highest sales in November (352k) and remains high in December.

Hence, the sale is highest in November and lowest in February.

### Analyzing Sales based on product categories

```python
sales_by_category = df.groupby('Category')['Sales'].sum().reset_index()
sales_by_category.sort_values(by='Sales', ascending=False, inplace=True)
sales_by_category.reset_index(drop=True, inplace=True)
sales_by_category
```

There are only 3 categories in the dataset. `Technology` category has the highest sales of 836,154, followed by `Furniture` and `Office Supplied` categories.

Creating a pie chart to show the sales of each category

```python
fig = px.pie(sales_by_category, values='Sales', names='Category', title='Sales by Category')
fig.show()
```

As we can see from the pie chart, the `Technology` category consists of 36.4%, followed by `Furniture` and `Office Supplied` categories with sales of 32.3% and 31.3% respectively.

Creating a same pie chart but in a fancy way with a hole in the middle and changing the color palette.

```python
fig = px.pie(sales_by_category, 
values='Sales', 
names='Category', 
title='Sales by Category',
color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(hole=.4, 
textinfo='label+percent', 
textposition='inside',
marker=dict(colors=px.colors.qualitative.Pastel))

fig.update_layout(title_font_size=20)

fig.show()
```

### Analyzing Sales by Sub Category

```python
sales_by_subcategory = df.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subcategory.sort_values(by='Sales', ascending=False, inplace=True)
sales_by_subcategory.reset_index(drop=True, inplace=True)
sales_by_subcategory
```

There are 16 sub-categories in the dataset. `Phones` sub-category has the highest sales of 330,007, followed by `Chairs` and `Storage`. `Fasteners` sub-category has the lowest sales of 3024 only.

Creating a bar chart to show the sales of each sub-category

```python
fig = px.bar(sales_by_subcategory, 
x='Sub-Category', 
y='Sales', 
title='Sales by Sub-Category')
fig.show()
```

```python
print("Phones % of sales: ", float(sales_by_subcategory[sales_by_subcategory["Sub-Category"] == "Phones"]["Sales"]/sales_by_subcategory["Sales"].sum() * 100))
print("Fasteners % of sales: ", float(sales_by_subcategory[sales_by_subcategory["Sub-Category"] == "Fasteners"]["Sales"]/sales_by_subcategory["Sales"].sum() * 100))
```

```python
print("There are 16 sub-categories in the dataset.")
print("Phones sub-category has the highest sales which comes under", df[df["Sub-Category"] == "Phones"]["Category"].values[0], "category", "with sales of", sales_by_subcategory[sales_by_subcategory["Sub-Category"] == "Phones"]["Sales"].values[0])
print("Fasteners sub-category has the lowest sales which comes under", df[df["Sub-Category"] == "Fasteners"]["Category"].values[0], "category", "with sales of", sales_by_subcategory[sales_by_subcategory["Sub-Category"] == "Fasteners"]["Sales"].values[0])
```

### Analyzing the Monthly profit by category

```python
profit_by_month = df.groupby("Order_Month")["Profit"].sum().reset_index()
profit_by_month.sort_values(by="Profit", ascending=False, inplace=True)
profit_by_month.reset_index(drop=True, inplace=True)
profit_by_month
```

As we can see, even though `November` has the highest sales, `December` has the highest profit, followed by `September` and then `November`. However, `January` and `February` have the lowest profit because they have the lowest sales.

Creating a bar chart and line chart to show the profit by month

```python
fig = px.bar(profit_by_month, x="Order_Month", y="Profit", title="Profit by Month")
fig.show()
```

```python
fig = px.line(profit_by_month.sort_values(by="Order_Month", ascending=True), x="Order_Month", y="Profit", title="Profit by Month")
fig.show()
```

We can clearly see that `December` has the highest profit and `January` has the lowest profit. The profit increased in `March` but decreased rapidly in `April`. The profit then increased agin and decreased. It started to steadily increase from `August` to `December`.

Calculating the average profit per month

```python
profit_by_month_avg = df.groupby("Order_Month")["Profit"].mean().reset_index()
profit_by_month_avg.sort_values(by="Profit", ascending=False, inplace=True)
profit_by_month_avg.reset_index(drop=True, inplace=True)
profit_by_month_avg
```

### Analyzing Profit by Category and Sub-Category

```python
profit_by_category = df.groupby('Category')['Profit'].sum().reset_index()
profit_by_category.sort_values(by='Profit', ascending=False, inplace=True)
profit_by_category.reset_index(drop=True, inplace=True)
profit_by_category
```

As we can see, `Technology` has the highest profit and the highest sales. However, even though `Furniture` has little higher sales than `Office Supplies`, `Office Supplies` had much larger profit margin than `Furniture`. Hence, company should focus on `Office Supplies` to increase profit margin even though it has lower sales than `Furniture`.

Creating a pie chart to show the profit margin of each category

```python
fig = px.pie(profit_by_category, 
values='Profit', 
names='Category', 
title='Profit by Category',
color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_traces(hole=.4, 
textinfo='label+percent', 
textposition='inside',
marker=dict(colors=px.colors.qualitative.Pastel))

fig.update_layout(title_font_size=20)

fig.show()
```

As we can see, `Technology` contributes 50.8% of the total profit with a sales of 36.4%, while `Furniture` contributes 6.44% of the total profit with a sales of 32.3%, `Office Supplies` contributes 42.8% of the total profit with a sales of 31.3%.

Now, calculating the profit by sub category

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_by_subcategory = profit_by_subcategory.sort_values(by='Profit', ascending=False)
profit_by_subcategory.reset_index(drop=True, inplace=True)
profit_by_subcategory
```

`Copiers` sub-category has the highest profit of 55,618, followed by `Phones` which has the highest sales and `Accessories`. However, `Tables` sub-category has the lowest profit of -17,725, followed by `Bookcases` and `Supplies`. This could be because this subcategory are furtinures which require a high cost of materials and labor to make them which make them lose money. However, `Fasteners` which had the lowest sales has a positive profit of 950 which is better than others where the sub category is lossing money.

```python
print("Copiers with highest profit come under", df[df["Sub-Category"] == "Copiers"]["Category"].values[0])
print("Tables with lowest profit come under", df[df["Sub-Category"] == "Tables"]["Category"].values[0])
```

Creating a bar chart to show the profit of each sub-category

```python
fig = px.bar(profit_by_subcategory, x='Sub-Category', y='Profit', title='Profit by Sub-Category')
fig.show()
```

### Analyzing sales and profit by customer segment

```python
customer_segment = df.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
customer_segment.sort_values(by='Profit', ascending=False, inplace=True)
customer_segment.reset_index(drop=True, inplace=True)
customer_segment
```

There are 3 customer segments. `Consumer` segment has the highest profit and sales of 134,119 and 1,161,401, followed by `Corporate` and `Home Office`.


Creating a double bar chart to compare sales and profit of each customer segment.

```python
fig = go.Figure()

fig.add_trace(go.Bar(x=customer_segment['Segment'], 
y=customer_segment['Sales'], 
name='Sales', 
marker_color = colors.qualitative.Pastel[0]))

fig.add_trace(go.Bar(x=customer_segment['Segment'], 
y=customer_segment['Profit'], 
name='Profit', 
marker_color= colors.qualitative.Pastel[1]))

fig.update_layout(
    title='Sales and Profit by Customer Segment',
    xaxis_title='Customer Segment',
    yaxis_title='Amount',
    barmode='group'
)

fig.show()
```

As we can clearly see from the bar chart, Consumer has the highest sales and profit, followed by Corporate and Home Office.

---
### Analyzing sales to profit ratio

```python
customer_segment['Sales to Profit Ratio'] = customer_segment['Sales'] / customer_segment['Profit']
customer_segment[["Segment", "Sales to Profit Ratio"]].sort_values(by='Sales to Profit Ratio', ascending=False)
```

`Consumer` segment has the highest sales to profit ratio of 8.66, followed by `Corporate` and `Home Office` segments. This indicates that the `Consumer` segment generates the most profit relative to its sales.

Visualize the sales to profit ratio for each segment using a bar chart.

```python
fig = px.bar(customer_segment, x='Segment', y='Sales to Profit Ratio', title='Sales to Profit Ratio by Segment')
fig.show()
```
