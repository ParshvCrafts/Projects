## Plotly Libraries

**plotly.graph_objects:** 
This is a low level interface to figures, traces and layout. The Plotly graph objects module provides an automatically generated hierarchy of classes ( figures, traces, and layout) called graph objects. These graph objects represent figures with a top-level class plotly.graph_objects.Figure.

**plotly.express:** 
Plotly express is a high-level wrapper for Plotly. It uses graph objects internally.

I used plotly_graph_objects to plot line and scatter plots

```python
# Importing required libraries
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

## Scatter Plot: 
A scatter plot shows the relationship between 2 variables on the x and y-axis. The data points here appear scattered when plotted on a two-dimensional plane. Using scatter plots, I created exciting visualizations to express relationship of income vs age


```python
# income vs age of people in a scatter plot
age_array=np.random.randint(25,55,60)
income_array=np.random.randint(300000,700000,3000000)
```

```python
# creating  an empty figure using go.Figure()
fig=go.Figure()
fig
```

```python
# creating a scatter plot by using the add_trace function and use the go.scatter() function within it
# In go.Scatter I defined the x-axis data,y-axis data and define the mode as markers with color of the marker as blue
fig.add_trace(go.Scatter(x=age_array, y=income_array, mode='markers', marker=dict(color='blue')))
```

Using the update_layout function to update the title and labels.


```python
fig.update_layout(title='Economic Survey', xaxis_title='Age', yaxis_title='Income')
fig.show()
```

#### Analysis:
     
- From the above plot I found that the Income of a person is not correlated with age. As the age increases, the income may or not decrease in this randomly generated dataset.


## Line Plot: 
A line plot shows information that changes continuously with time. Here the data points are connected by straight lines. Line plots are also plotted on a two dimensional plane like scatter plots. Using line plots, I created a random sample visualization of sales of bicycles.


```python
# sales of bicycles from Jan to August last year using a line chart
# Defining an array containing numberofbicyclessold  
numberofbicyclessold_array=[50,100,40,150,160,70,60,45]
# Defining an array containing months
months_array=["Jan","Feb","Mar","April","May","June","July","August"]
```

```python
fig=go.Figure()
# creating a line plot by using the add_trace function and use the go.scatter() function within it
# In go.Scatter I defined the x-axis data,y-axis data and defined the mode as lines with color of the marker as green
fig.add_trace(go.Scatter(x=months_array, y=numberofbicyclessold_array, mode='lines', marker=dict(color='green')))
```

```python
fig.update_layout(title='Bicycle Sales', xaxis_title='Months', yaxis_title='Number of Bicycles Sold')
fig.show()
```

#### Analysis:
     
- From the above plot, I found that the sales is the highest in the month of May (160 bicycles) and then there is a decline in sales.


  
## Bar Plot: 
A bar plot represents categorical data in rectangular bars. Each category is defined on one axis, and the value counts for this category are represented on another axis. Bar charts are generally used to compare values. To visualize, I created a bar plot of average pass percentage of classes.


```python
# average pass percentage of classes from grade 6 to grade 10

# Defining an array containing scores of students 
score_array=[80,90,56,88,95]
# Defining an array containing Grade names  
grade_array=['Grade 6','Grade 7','Grade 8','Grade 9','Grade 10']
```

In plotly express, setting axis values and the title within the same function call `px.<graphtype>(x=<xaxis value source>,y=<y-axis value source>,title=<appropriate title as a string>)`. In the below code I used `px.bar( x=grade_array, y=score_array, title='Pass Percentage of Classes')`.


```python
# Using plotly express bar chart function px.bar.Provide input data, x and y axis variable, and title of the chart.
fig = px.bar( x=grade_array, y=score_array, title='Pass Percentage of Classes') 
fig.show()
```

#### Analysis:
     
- From the above plot, I found that Grade 8 has the lowest pass percentage (56%) and Grade 10 has the highest pass percentage (95%) 



## Histogram: 
 A histogram is used to represent continuous data in the form of bar. Each bar has discrete values in bar graphs, whereas in histograms, we have bars representing a range of values. Histograms show frequency distributions. Visualizing distribution of heights.


```python
# distribution of heights of 200 people using a histogram
import numpy as np
# heights which are 160 and the standard deviation is 11
heights_array = np.random.normal(160, 11, 200)
## Use plotly express histogram chart function px.histogram.Provide input data x to the histogram
fig = px.histogram(x=heights_array,title="Distribution of Heights")
fig.show()
```

#### Analysis:
- From this I can analyze that there is around 1 person who is at the height of 130cm and 34 people at the height of 160 cm


## Bubble Plot: 
A bubble plot is used to show the relationship between 3 or more variables. It is an extension of a scatter plot. Bubble plots are ideal for visualizing:

  * Global Economic position of Industries
  * Impact of viruses on Diseases


```python
# crime statistics of US cities with a bubble chart

#Creating a dictionary having city,numberofcrimes and year as 3 keys
crime_details = {
    'City' : ['Chicago', 'Chicago', 'Austin', 'Austin','Seattle','Seattle'],
    'Numberofcrimes' : [1000, 1200, 400, 700,350,1500],
    'Year' : ['2007', '2008', '2007', '2008','2007','2008'],
}
  
# creating a Dataframe object with the dictionary
df = pd.DataFrame(crime_details)
df
```

```python
## Grouping the number of crimes by city and find the total number of crimes per city
bub_data = df.groupby('City')['Numberofcrimes'].sum().reset_index()
bub_data
```

```python
## Bubble chart using px.scatter function with x ,y and size varibles defined.Title defined as Crime Statistics
fig = px.scatter(bub_data, x="City", y="Numberofcrimes", size="Numberofcrimes",
                 hover_name="City", title='Crime Statistics', size_max=60)
fig.show()
```

#### Analysis:
- The size of the bubble in the bubble chart indicates that Chicago has the highest crime rate (2200 crimes) when compared with the other 2 cities.


 ## Pie Plot: 
 A pie plot is a circle chart mainly used to represent proportion of part of given data with respect to the whole data. Each slice represents a proportion and on total of the proportion becomes a whole. Bar plots are useful in visualizing:
 
 * Sales turnover percentatge with respect to different products
 * Monthly expenditure of a Family


```python
## Monthly expenditure of a family
exp_percent= [20, 50, 10,8,12]
house_holdcategories = ['Grocery', 'Rent', 'School Fees','Transport','Savings']
```

```python
# Using px.pie function to create the chart. 
# Values parameter set values associated to the sector. 'exp_percent' feature is passed to it.
# labels for the sector are passed to the `house hold categoris` parameter.
fig = px.pie(values=exp_percent, names=house_holdcategories, title='Household Expenditure')
fig.show()
```

#### Analysis
- From this pie chart I found that the family expenditure is maximum for rent (50%), then grocery (20%) and savings(12%).


## Sunburst Charts: 
Sunburst charts represent hierarchial data in the form of concentric circles. Here the innermost circle is the root node which defines the parent, and then the outer rings move down the hierarchy from the centre. They are also called radial charts. Used for:

* Worldwide mobile Sales where we can drill down as follows:   
    * innermost circle represents total sales  
    * first outer circle represents continentwise sales
    * second outer circle represents countrywise sales within each continent
    
    
* Disease outbreak hierarchy


* Real Estate Industrial chain


```python
# Creating a dictionary having a set of people represented by a character array and the parents of these characters represented in another
# array and the values are the values associated to the vectors.
data = dict(
    character=["Eve", "Cain", "Seth", "Enos", "Noam", "Abel", "Awan", "Enoch", "Azura"],
    parent=["", "Eve", "Eve", "Seth", "Seth", "Eve", "Eve", "Awan", "Eve" ],
    value=[10, 14, 12, 10, 2, 6, 6, 4, 4])

fig = px.sunburst(
    data,
    names='character',
    parents='parent',
    values='value',
    title="Family chart"
)
fig.show()
```

#### Analysis

- The innermost circle **Eve** represents the parent and the second outer circle represents his childrent **Cain**,**Seth** and so on. Further the outermost circle represents his grandchildren **Enoch** and **Enos**


## Appling Plotly to visualize an Airline Dataset

The Reporting Carrier On-Time Performance Dataset contains information on approximately 200 million domestic US flights reported to the United States Bureau of Transportation Statistics. The dataset contains basic information about each flight (such as date, time, departure airport, arrival airport) and, if applicable, the amount of time the flight was delayed and information about the reason for the delay. This dataset can be used to predict the likelihood of a flight arriving on time.

Preview data, dataset metadata, and data glossary [here.](https://dax-cdn.cdn.appdomain.cloud/dax-airline/1.0.1/data-preview/index.html)


```python
# Reading the airline data into pandas dataframe
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/airline_data.csv'

airline_data =  pd.read_csv(URL,
                            encoding = "ISO-8859-1",
                            dtype={'Div1Airport': str, 'Div1TailNum': str, 
                                   'Div2Airport': str, 'Div2TailNum': str})

airline_data.head()
```

```python
# Shape of the data
airline_data.shape
```

```python
# Randomly sample 500 data points. Setting the random state to be 42 to get same result.
data = airline_data.sample(n=500, random_state=42)
```

```python
# Geting the shape of the trimmed data
data.shape
```

Visually capturing details such as

* Departure time changes with respect to airport distance.

* Average Flight Delay time over the months

* Comparing number of flights in each destination state

* Number of  flights per reporting airline

* Distrubution of arrival delay

* Proportion of distance group by month (month indicated by numbers)

* Hierarchical view in othe order of month and destination state holding value of number of flights


## Scatter Plot


I used a scatter plot to represent departure time changes with respect to airport distance


```python
fig=go.Figure()
# creatung a scatter plot by using the add_trace function and use the go.scatter() function within it
fig.add_trace(go.Scatter(x=data['Distance'], y=data['DepTime'], mode='markers', marker=dict(color='red')))
fig.update_layout(title='Distance vs Departure Time', xaxis_title='Distance', yaxis_title='DepTime')
fig.show()
```

#### Analysis:

- It can be inferred that there are more flights round the clock for shorter distances. However, for longer distance there are limited flights through the day.


## Line Plot


I used a line plot to extract average monthly arrival delay time and see how it changes over the year.

```python
# Grouping the data by Month and compute average over arrival delay time.
line_data = data.groupby('Month')['ArrDelay'].mean().reset_index()
line_data
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(x=line_data['Month'], y=line_data['ArrDelay'], mode='lines', marker=dict(color='green')))
fig.update_layout(title='Month vs Average Flight Delay Time', xaxis_title='Month', yaxis_title='ArrDelay')
fig.show()
```

#### Analysis:

- It is found that in the month of `June` the average monthly delay time (17.3 minutes) is the maximum, whereas in `November`, the planes arrive early by 4 minutes on average.


## Bar Chart



I used a bar chart to extract number of flights from a specific airline that goes to a destination


```python
# Grouping the data by destination state and reporting airline. Compute total number of flights in each combination
bar_data = data.groupby('DestState')['Flights'].sum().reset_index()
bar_data
```

```python
# Using plotly express bar chart function px.bar. 
fig = px.bar(bar_data, x="DestState", y="Flights", title='Total number of flights to the destination state split by reporting airline') 
fig.show()
```

#### Analysis:

- It is found that maximum flights are to destination state `CA` (California) which is 68 flighst and there are only 1 flight to 9 destination states `IA` (Iowa), `ID` (Idaho), `KS` (Kansas), `NH` (New Hampshire), `NM` (New Mexico), `RI`, `SC` (South Carolina), and `VI` (United States Virgin Islands).

## Histogram


Reresenting the distribution of arrival delay using a histogram


```python
data["ArrDelay"].isna().sum()
```

```python
# Seting missing values to 0
data['ArrDelay'] = data['ArrDelay'].fillna(0)
```

## Bubble Chart


I used a bubble plot to represent number of flights as per reporting airline

```python
# Group the data by reporting airline and get number of flights
bub_data = data.groupby('Reporting_Airline')['Flights'].sum().reset_index()
bub_data
```

```python
# Bubble chart using px.scatter function with x ,y and size variables defined.Title defined as Reporting Airline vs Number of Flights
fig = px.scatter(bub_data, x="Reporting_Airline", y="Flights", size="Flights",
                 hover_name="Reporting_Airline", title='Reporting Airline vs Number of Flights', size_max=60)
fig.show()
```

#### Analysis

- It is found that the reporting airline **WN** (Southwest Airlines) has the highest number with 86 flights, followed by airline **DL** (Delta Air Lines) with 66 flights and **AA** (American Airlines) with 57 flights

## Pie Chart


Representing the proportion of Flights by Distance Group (Flights indicated by numbers)

```python
# Using px.pie function to create the chart.  
# Values parameter will set values associated to the sector. 'Flights' feature is passed to it.
fig = px.pie(data, values='Flights', names='DistanceGroup', title='Flight propotion by Distance Group')
fig.show()
```

#### Analysis

- It is found that `Distance group 2` has the highest flight proportion with 136 flights which is 27.2%, followed by `Distance group 3` with 91 flights (18.2%) and `Distance group 1` with 90 flights (18%).


## SunBurst Charts


Representing the hierarchical view in othe order of month and destination state holding value of number of flights

```python
## Defining path as Month and DestStateName and values as Flights.
fig = px.sunburst(data, path=['Month', 'DestStateName'], values='Flights',title='Flight Distribution Hierarchy')
fig.show()
```

#### Analysis

- **Month** numbers present in the innermost concentric circle is the root and for each month, check the **number of flights** for the different **destination states** under it.
