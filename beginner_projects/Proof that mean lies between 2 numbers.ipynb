```python
"""
Mean Myth Buster
Test whether the mean of numbers always lies between the min and max.

Author: Parshv Patel | June 2025
"""

import random
from statistics import mean

def test_mean_within_bounds(trials: int, values_per_trial: int = 2, log_to_file: bool = False):
    results = []
    
    for trial in range(1, trials + 1):
        nums = [random.randint(1, 100) for _ in range(values_per_trial)]
        avg = mean(nums)
        result = min(nums) <= avg <= max(nums)
        results.append((trial, nums, round(avg, 2), result))
    
    true_count = sum(1 for _, _, _, r in results if r)
    
    print(f"\nOut of {trials} trials, mean was within bounds in {true_count} cases.")
    if true_count == trials:
        print("Math never lies: The mean was always between min and max!")
    else:
        print("⚠️ Surprise! There were some unexpected cases.")

    print("\nSample Trials:")
    for trial, nums, avg, result in results[:min(5, trials)]:
        print(f"Trial {trial}: {nums} -> mean = {avg} -> within bounds? {result}")

    if log_to_file:
        with open("mean_test_log.txt", "w") as file:
            for trial, nums, avg, result in results:
                file.write(f"Trial {trial}: {nums} → mean = {avg} → within bounds? {result}\n")
        print("\nFull results saved to 'mean_test_log.txt'.")

if __name__ == "__main__":
    try:
        trials = int(input("How many times should we test the mean logic?"))
        val_count = int(input("How many numbers per trial (e.g. 2 or more)?"))
        save = input("Save results to file? (y/n)").strip().lower() == 'y'
        test_mean_within_bounds(trials, val_count, save)
    except ValueError:
        print("⚠️ Please enter valid integer values.")
```
